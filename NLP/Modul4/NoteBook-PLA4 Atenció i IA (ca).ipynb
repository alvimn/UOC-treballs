# Módulo 4: Implementació del Deep Learning al Processament del Llenguatge Natural

## Scripts de models del llenguatge aplicats a la intel·ligència artificial

En aquest notebook il·lustrem l'aplicació d'un model del llenguatge a una tasca de la intel·ligència artificial. La tasca consisteix en determinar si una frase pot inferir-se de la frase que la precedeix en termes de similitud semántica

### Instal·lació del paquet <i>semantic-text-similarity</i>

```python
pip install semantic-text-similarity
```

## Carregar el model

Definir el model sobre el qual es calcula la similitud. Es el model <i>WebBertSimilarity</i>, fet a partir de pàgines web

```python
from semantic_text_similarity.models import WebBertSimilarity
```

```python
web_model = WebBertSimilarity(device='cpu', batch_size=10) #Establecer que el dispositivo es un CPU
```

### Predicció

```python
print(web_model.predict([("She won an olympic gold medal","The woman is an olympic champion")]))
```

```python
print(web_model.predict([("She won an olympic gold medal","The man is a teacher")]))
```
