# Traducción automática con Deep Learning

Este notebook muestra cómo hacer un prototipo de traductor automático alemán-inglès con un métode de <i>Deep Learning</i>. El traductor aprende de unos pares de frases en alemán con sus respectivas traducciones humanas al inglès

Este notebook está inspirado en el post <a href="https://www.analyticsvidhya.com/blog/2019/01/neural-machine-translation-keras/">A Must-Read NLP Tutorial on Neural Machine Translation – The Technique Powering Google Translate</i></a>, de Prateek Joshi

## Cargar el dataset de pares original-traducción

```python
from numpy import array, argmax, random, take

def read_text(filename):
        file = open(filename, mode='rt', encoding='utf-8')
        # read all text
        text = file.read()
        file.close()
        return text

# Convertir el texto a pares de oraciones inglés-alemán
def to_lines(text):
    sents = text.strip().split('\n')
    #En cada línea, tomar los dos primeros elementos separados por un tabulador. El primer elemento es la frase
    #en inglés y el segundo elemento es su equivalente al alemán
    sents = [i.split('\t')[:2] for i in sents]
    return sents    

# El fichero deu.txt contiene el dataset
data = read_text("./SOURCE/deu.txt")
deu_eng = to_lines(data)
deu_eng = array(deu_eng)

print(deu_eng)
```

## Preparar los pares original-traducción de entrenamiento

Tomamos los primeros 50.000 pares

```python
deu_eng = deu_eng[:50000,:]

print(deu_eng)
```

### Preprocesamiento

Limpieza de signos de puntuación y conversión de los textos a minúscula

```python
import string

# Quitamos signos de puntuación
deu_eng[:,0] = [s.translate(str.maketrans('', '', string.punctuation)) for s in deu_eng[:,0]]
deu_eng[:,1] = [s.translate(str.maketrans('', '', string.punctuation)) for s in deu_eng[:,1]]

# Convertimos los textos a minúsculas
for i in range(len(deu_eng)):
    deu_eng[i,0] = deu_eng[i,0].lower()
    deu_eng[i,1] = deu_eng[i,1].lower()

print(deu_eng)
```

## Codificación de los datos

La codificación consiste en lo siguiente:
<ul>
<li> Representación numérica de las palabras
<li> Codificación de las secuencias
</ul>

Hay que transformar las oraciones a vectores de la misma longitud. Idealmente cada dimensión corresponde a una palabra de la oración. Ahora bien, como que no todas las oraciones tienen la misma longitud, se genera una secuencia numérica cuya longitud corresponde a la oración más larga. Cuando se codifica la frase, las dimensiones más allá de su longitud se llenan de 0 (<i>padding</i>).

### Representación numérica de las palabras (tokenización)

La tokenización se ocupa de crear un vocabulario donde las palabras de las oraciones se representan como un número. Este número es el índice de posición en el vocabulario. El ínidice de posición depende del orden de las palabras por su frecuencia.

```python
from keras.preprocessing.text import Tokenizer

def tokenization(sentences):
    tokenizer = Tokenizer()
    tokenizer.fit_on_texts(sentences)
    return tokenizer
```

#### Tokenización de las frases en inglés y en alemán

```python
#Tokenizar las frases en inglés
eng_tokenizer = tokenization(deu_eng[:, 0])

deu_tokenizer = tokenization(deu_eng[:, 1])

# Ver el vocabulario del alemán (Diccionario cuyas keys son las palabras y sus valores son sus índices)
print(deu_tokenizer.word_index)
```

```python
#Tokenizar las frases en inglés
eng_tokenizer = tokenization(deu_eng[:, 0])
#Ob
eng_vocab_size = len(eng_tokenizer.word_index) + 1

eng_length = 8
print('English Vocabulary Size: %d' % eng_vocab_size)

deu_tokenizer = tokenization(deu_eng[:, 1])
deu_vocab_size = len(deu_tokenizer.word_index) + 1

deu_length = 8
print('Deutch Vocabulary Size: %d' % deu_vocab_size)
```

### Codificación de las secuencias

Distinguimos primero las secuencias- es decir, las frases- para el entrenamiento (training) y las secuencias destinadas para el test

```python
from sklearn.model_selection import train_test_split

train, test = train_test_split(deu_eng, test_size=0.2, random_state = 12)
```

Para codificar las secuencias en una lengua <i>L</i> necesitamos:
<ul>
<li> El vocabulario de la lengua
<li> La longitud máxima que puede tener una frase
<li> Las frases a codificar
</ul>

Asumimos que hemos hecho previamente el cálculo de la longitud máxima de las oraciones en inglés y alemán (11 en alemán i 8 en inglés).

La codificación consiste en crear un vector con un número de dimensiones predefinido. Cada dimensión del vector es el índice de una palabra según el vocabulario. Como que los vectores deben tener el mismo número de dimensiones, se toma la longitud máxima de una frase como el número de dimensiones del vector. Si la frase a codificar tiene una longitud menor, se llenan las dimensiones sobrantes con un 0 (padding). La longitud máxima establecida es 8

```python
eng_length = 8
deu_length = 8
```

```python
from keras.preprocessing.sequence import pad_sequences

def encode_sequences(tokenizer, length, lines):
    # Codificar las secuencias con los índices de las palabras
    seq = tokenizer.texts_to_sequences(lines)
    # Hacer el padding
    seq = pad_sequences(seq, maxlen=length, padding='post')
    return seq
```

#### Codificación de las secuencias de los originales y de las traducciones del training

```python
trainX = encode_sequences(deu_tokenizer, deu_length, train[:, 1])
trainY = encode_sequences(eng_tokenizer, eng_length, train[:, 0])
```

```python
# Ver las frases en alemán del training
print(train[:,1])
```

```python
# Ver las misma frases codificadas
print(trainX)
```

#### Codificación de las secuencias de los originales y de las traducciones del test

```python
testX = encode_sequences(deu_tokenizer, deu_length, test[:, 1])
testY = encode_sequences(eng_tokenizer, eng_length, test[:, 0])
```

# Creación del modelo de traducción

El modelo de traducción tiene una arquitectura Seq2Seq. Esta arquitectura tiene dos componentes:

<ul>
<li>Codificador (encoder)
<li>Descodificador (decoder)
</ul>

Para el <i>codificador</i> se utilizará una capa embedding y una capa LSTM. La capa embedding es donde se introducen las representaciones vectoriales de las palabras en la lengua origen

Para el <i>descodificador</i> se utilizará una capa LSTM seguida de una capa densa para predecir la cadena de palabras más probable de ser la oración destino.


## Capa embedding

### Ejemplo de creación de una capa embedding

Imaginemos que queremos hacer la capa embedding para las siguientes frases:<br>

<i><b>Generamos una capa embedding</i></b><br>
<i><b>No generamos una capa embedding</i></b>

Primeros tokenizamos y creamos el vocabulario

```python
texts = ['generamos una capa embedding', 'no generamos una capa embedding']

t_tokenizer = tokenization(texts)
t_vocab_size = len(t_tokenizer.word_index) + 1

print("Vocabulary")
print(t_tokenizer.word_index)
print("Vocabulary size")
vocab_size = len(t_tokenizer.word_index) + 1
print(vocab_size)
```

Establecemos la longitud máxima de las secuencias y la dimensión de los vectores

```python
max_text_length = 5 # Longitud de la frase más larga

embedding_vec_length = 3
```

Codificamos las secuencias

```python
enc_seq = encode_sequences(t_tokenizer, max_text_length, texts)

print(enc_seq)
```

```python
from keras.models import Sequential
from keras.layers import Dense, LSTM, Embedding, RepeatVector

# Definimos un modelo secuencial sólo con la capa embedding:

def define_model(vocab_size, vec_length, max_text_length):
# El modelo es secuencial
    model = Sequential()
# Añadir la capa embedding:
# La capa embedding se inicializa con pesos aleatorios y aprende una representación vectorial (embeddings) 
# para todas las palabras de las frases. Los embeddings podrián ser también vectores Word2Vec o Glove precalculados

# Es necesario tener: 
# el tamaño del vocabulario de los textos (vocab_size)
# la dimensión de los vectores en los que las palabras serán embedded (embedding_vec_length).
# la longitud máxima de las secuencias de input
    model.add(Embedding(vocab_size, embedding_vec_length, input_length=max_text_length, mask_zero=True))
#Predecimos los embeddings según el modelo
    output_array = model.predict(enc_seq)
#Ver el output del modelo con la capa embedding
    print("Embeddings")
    print(output_array)
    return model

model = define_model(vocab_size, embedding_vec_length, max_text_length)
```

### Creación de la capa embedding en el encoder del traductor

```python
from keras.models import Sequential
from keras.layers import Dense, LSTM, Embedding, RepeatVector

embedding_vec_length = 512

def define_model(in_vocab_size, embedding_vec_length, max_text_length):
    mt_model = Sequential()
    mt_model.add(Embedding(in_vocab_size, embedding_vec_length, input_length = max_text_length, mask_zero=True))
    output_array = mt_model.predict(trainX)
    print(output_array)
    return mt_model

deu_vocab_size = len(deu_tokenizer.word_index) + 1

#Definimos el modelo con una capa de embedding para las frases de la lengua origen (alemán)
mt_model = define_model(deu_vocab_size, embedding_vec_length, deu_length)
```

## Encoder del traductor (Embedding layer + LSTM layer)

```python
from keras.models import Sequential
from keras.layers import Dense, LSTM, Embedding, RepeatVector

def define_model(in_vocab_size, embedding_vec_length, max_text_length):
    mt_model = Sequential()
    #Embedding Layer
    mt_model.add(Embedding(in_vocab_size, embedding_vec_length, input_length = max_text_length, mask_zero=True))
    #Añadimos el LSTM layer, con el número de unidades ('units').
    #'Units' es el número de 'hidden units' en cada intervalo de tiempo en la red LSTM
    mt_model.add(LSTM(units))
    return mt_model

units = 512

#La longitud de los vectores corresponde al número de units del LSTM

embedding_vec_length = units

mt_model = define_model(deu_vocab_size, embedding_vec_length, deu_length)
```

## Decoder del traductor

```python
from keras.models import Sequential
from keras.layers import Dense, LSTM, Embedding, RepeatVector

def define_model(in_vocab_size, embedding_vec_length, max_text_length, out_timesteps, out_vocab_size):
    mt_model = Sequential()
    mt_model.add(Embedding(in_vocab_size, embedding_vec_length, input_length = max_text_length, mask_zero=True))
    mt_model.add(LSTM(units))
    #Descodificador:
    #Repetición del último output tantas veces como la longitud máxima de la lengua destino (eng_length).
    #Fíjese que esta longitud máxima corresponde a estados de tiempo del output (out_timesteps)
    #https://stackoverflow.com/questions/51749404/how-to-connect-lstm-layers-in-keras-repeatvector-or-return-sequence-true
    mt_model.add(RepeatVector(out_timesteps))
    #Añadir la capa LSTM con el mismo número de units que el LSTM del codificador. Establecemos que queremos
    #obtener la secuencia completa del output; esto es, la traducción de la frase entera.
    mt_model.add(LSTM(units, return_sequences=True))
    #Añadir la capa donde se representa la distribución de las palabras del vocabulario de la lengua destino 
    #según la probabilidad de aparición en la secuencia destino. Así podemos obtener la secuencia más probable
    #de ser la traducción de la frase origen
    mt_model.add(Dense(out_vocab_size, activation='softmax'))
    return mt_model

units = 512

#La longitud de los vectores corresponde al número de units del LSTM

embedding_vec_length = units

#Tamaño del vocabulario de la lengua destino
eng_vocab_size = len(eng_tokenizer.word_index) + 1

mt_model = define_model(deu_vocab_size, embedding_vec_length, deu_length, eng_length, eng_vocab_size)
```

# Compilación y optimización del modelo

Compilamos y optimizamos el modelo con el <i>RMSprop optimizer</i><br>

Según Prateek Joshi a <i>A Must-Read NLP Tutorial on Neural Machine Translation – The Technique Powering Google Translate</i><br>

<i>We have used ‘sparse_categorical_crossentropy‘ as the loss function. 
This is because the function allows us to use the target sequence as is, instead of the one-hot encoded format.
One-hot encoding the target sequences using such a huge vocabulary might consume our system’s entire memory.</i>

```python
from keras import optimizers

rms = optimizers.RMSprop(lr=0.001)
mt_model.compile(optimizer=rms, loss='sparse_categorical_crossentropy')
```

# Training

<b>¡¡¡ATENCIÓN!!!: El training puede durar horas. Tenedlo en cuenta porque el entrenamiento de un traductor automático sencillo como éste puede llegar a tardar hasta 8 horas. Por tanto, podéis ir a dormir y dejadlo trabajar.

```python
from keras.callbacks import ModelCheckpoint

filename = 'model_ta_deu_en' #Fichero donde guardaremos el modelo para hacer las posteriores predicciones

#La función ModelCheckpoint() guarda el modelo con la pérdida de validación más baja.

checkpoint = ModelCheckpoint(filename, monitor='val_loss', verbose=1, save_best_only=True, mode='min')

#Entrenamiento del modelo. Se realiza con 30 epoch con un tamaño de batch de 512, con un reparto de validación del
#20%; esto es, 80% se destina al entrenamiento en sí, y el 20% restante a su validación.

mt_model.fit(trainX, trainY.reshape(trainY.shape[0], trainY.shape[1], 1),
                    epochs=30, batch_size=512, validation_split = 0.2,callbacks=[checkpoint], 
                    verbose=1)
```

# Predicción

Según el modelo, predecimos las secuencias numéricas target del test

```python
from keras.models import load_model

model = load_model('model_ta_deu_en')
preds = model.predict_classes(testX.reshape((testX.shape[0],testX.shape[1])))
print(preds)
```

Convertimos las secuencias de números a secuencias de palabras

```python
def get_word(n, tokenizer):
    for word, index in tokenizer.word_index.items():
        if index == n:
            return word
    return None

preds_text = []

for i in preds:
    temp = []
    for j in range(len(i)):
        #Obtener la palabra que corresponde al índice del vocabulario de la lengua destino
        t = get_word(i[j], eng_tokenizer)
        if j > 0:
            if (t == get_word(i[j-1], eng_tokenizer)) or (t == None):
                     temp.append('')
            else:
                     temp.append(t)
        else:
            if(t == None):
                temp.append('')
            else:
                temp.append(t) 

    preds_text.append(' '.join(temp))
```

Comparamos una muestra de 15 traducciones del sistema (predicted) con sus respectivas traducciones humanas (actual) del corpus de test. Como podéis ver, acierta en algunos casos, en otros se acerca y en otros se aleja.  Evidentemente, la mejora de resultados depende de la cantidad de datos de entrenamiento y de la capacidad de procesamiento de la máquina con cantidades ingentes de datos y más <i>epochs</i>

```python
import pandas as pd

pred_df = pd.DataFrame({'actual' : test[:,0], 'predicted' : preds_text})

pred_df.sample(15)
```
