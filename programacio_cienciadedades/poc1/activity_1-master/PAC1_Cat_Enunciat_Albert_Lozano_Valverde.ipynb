## Exercici 3

Utilitzant el *dataframe* obtingut en l'Exercici 1

creeu una funció tal que tingui com a paràmetres d'entrada:
* L'estructura de dades (dataframe obtingut en l'Exercici 1)
* Codi d'infracció
* Normativa

La funció ens retornarà un **boolean** que indiqui si la infracció que correspon al codi compleix amb la Normativa (legislació) de la columna "Normativa" (LSV o OCVV o RGC).

Si no s'introdueix cap valor en el paràmetre de normativa, que per defecte assigni el valor 'LSV'.

Per mostrar el resultat, retorna un text complet com aquest: "Per a la infracció amb codi 'codi_infraccio' amb data de vigència 'Data_inici_vigencia' compleix la normativa 'paràmetre de la normativa/no compleix la normativa". L'última part del text dependrà de si compleix o no la normativa que s'ha passat com a paràmetre.

<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px; ">SM</span> **(1 punt)**

<div style="background-color: #fcf2f2; border-color: #dfb5b4; border-left: 5px solid #dfb5b4; padding: 0.5em;">
<strong>Solució</strong>
</div>

```python

def valorador(df,cd,nor="LSV"): #posem "LSV" com a default de normativa
    try:
        filtrat=df[df["Codi"]==cd] 
        var=filtrat["Normativa"].iloc[0]==nor#Busquem el codi al df i ens quedem amb la columna normativa i  el primer valor que hauria de ser el unic, y comparem si es igual a la normativa donada
    except: 
        print("El codi "+ str(cd) + " no existeix" )# en cas que el codi no exiteixi t¡fem aixo
        return
    data=df[df["Codi"]==cd]["Data_Inici_Vigencia"].iloc[0]# extreimem la data d'inici de vigencia fen el mateix que abans
    if(var): #comprobem si complim la normativa per canviar el print final
        text="compleix la normativa"
    else:
        text="no compleix la normativa"
    print("Per a la infracció amb codi "+str(cd) + " amb data de vigència " +data+ " "+ text)
   
```

```python
valorador(df,1,"OCVV")
valorador(df,912,"OCVV")
valorador(df,912)
```

Imprimiu el resultat amb el format: "L'infracció amb codi {Infraccio_Codi} ha retirat al conductor {Punts_A_Retirar} per {Infraccio_Desc_Llarga_CA}".

Realitzeu la prova per al tipus d'infracció M i més de 3 punts.

```python
def infraccions(tp,pnt):
    temp=df[(df["Tipus_Infraccio"]==tp) & (df["Punts_A_Retirar"]>pnt)] #filtrem per les dos condicions donades
    for index, row in temp.iterrows(): #iterem i fem els prints corresponents
        print("L'infracció amb codi "+str(row["Codi"])+" ha retirat al conductor "+str(row["Punts_A_Retirar"])+" per "+str(row["Infraccio_Desc_Llarga_CA"]))
```

**Exercici: Gestió d'una llista d'infraccions pendents**

**Enunciat**:

Creeu un programa que mantingui una llista d'infraccions pendents de tramitar. Només contindrà codis d'infracció (1 columna).

* Comenceu amb una llista buida.
* Demaneu a l'usuari que afegeixi infraccions (codis d'infracció) a la llista.
* Permeteu a l'usuari treure una infracció de la llista si ja ha estat trametuda.
* Mostreu la llista d'infraccions pendents després de cada operació.

**Detalls**:

* El programa oferirà (demanarà a l'usuari) 3 opcions: Afegir infracció, eliminar infracció i sortir.
* Si escull l'opció Afegir, el programa ha de permetre afegir múltiples infraccions fins que l'usuari decideixi acabar.
* Si escull l'opció Eliminar, l'usuari pot eliminar una infracció especificant el seu codi. Si el codi d'infracció a eliminar no està a la llista, el programa ha de notificar a l'usuari.
* Si escull l'opció Sortir, el programa finalitza.

<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px;">SM</span> **(1.5 punts)**

```python
def app():
    infr=[]#la llista de infraccions
    a=True #comprobador si hem de parar 
    while a==True:
        print("\n Selecciona si vols:\n 1-afegir una infraccio \n 2-eliminar una infraccio \n 3-sortir")#mostrem per pantalla les tres opcions posibles
        x = input()#guardem l'imput
        if(x=="1"): #si vol afegir li preguntem quin vol afegir
            print("\n Has seleccionat afegir una infraccio, dona el codi de la infraccio a afegir") 
            y = input()
            if(y in infr):#si el valor ja esta a la llista ho comuniquem per pantalla per no tenir duplicatrs en cas que no ho estigui si que l'afegim utilitzant apend
                print("\n La infraccio ja estava a la llista \n")
            else: 
                infr.append(y)
        elif(x=="2"):#li comuniquem que ha seleccionat eliminar i li demanem el codi a eliminar
            print("\n Has seleccionat eliminar una infraccio, dona el codi de la infraccio a eliminar") 
            y = input()
            if(y in infr):#comprobem si el codi existeix, si existeix l'eliminem si no li comuniquem que el codi no esta a la llista
                infr.remove(y)
            else: 
                 print(" \n La infraccio no estava a la llista \n")
        elif(x=="3"):#canviem el valor del comprovant per sortir i li comuniquem que esta sortint
            print("\n Has seleccionat la opcio de sortir ")
            a=False
        else:# en cas que no posi un valor valid li diem quin son els valors valids i tornem
            print("\n Has de posar 1 2 o 3")
        print("\n La llista es \n")#mostrem per pantalla la llista
        print(infr)



```
