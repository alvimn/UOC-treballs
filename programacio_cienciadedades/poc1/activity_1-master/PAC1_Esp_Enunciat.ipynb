Es importante notar que estas etiquetas no indican el nivel de dificultad del ejercicio, sino únicamente la necesidad de consulta de documentación externa para su resolución. Además, recuerde que las **etiquetas son informativas**, pero puede consultar referencias externas siempre que lo desee (aunque no se indique explícitamente) o puede que pueda realizar una actividad sin consultar ningún tipo de documentación. ##  Enunciado
Los siguientes ejercicios se realizarán utlilizando el dataset denuncies_sancions.csv proporcionado juntamente con la PEC, y que contiene información sobre diferentes sanciones de tráfico en la ciudad de Barcelona, obtenido de la web de Open Data del Ayuntamiento de Barcelona. Cada fila representa una sanción.

Los campos del dataset son los siguientes:

   - Infraccio_codi: Código interno de la infracción.
   - Infraccio_Desc_Curta_CA: Descripción corta de la infracción en catalán.
   - Infraccio_Desc_Curta_ES: Descripción corta de la infracción en español.
   - Infraccio_Desc_Llarga_ES: Descripción detallada de la infracción en español.
   - Data_Inici_Vigencia: Fecha aplicación de la sanción.
   - Data_Fi_Vigencia: Fecha fin de la vigencia de la sanción (por defecto, siempre 31/12/9999)
   - Article: Código del artículo legal por el cual se aplica la sanción.
   - Normativa: Procedencia de la normativa del cual procede el artículo.
   - Import_Nominal_?: Importe de la sanción.
   - Tipus_Infraccio: Código del tipo de infracción.
   - Punts_A_Retirar: Puntos a retirar al conductor.
   - Autoritat: Autoridad que firma la sanción


**IMPORTANTE:** La carga del dataset se debe hacer usando **rutas relativas**, tenéis que cargar el fichero teniendo en cuenta que trabajáis (vuestro notebook se encuentra) en una carpeta que contiene la carpeta *data* y el archivo de datos correspondiente.
