## Exercici 1
Estudiarem la complexitat algorítmica en relació amb el temps d'execució de la funció següent, haureu de calcular la complexitat algorítmica Θ i justificar la vostra resposta. Podeu assumir que tots els paràmetres d'entrada tenen la mateixa mida.

<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px; ">SM</span> **(1 punto)**

```python
def algoritmicComplexity(L: List[int], m: List[int], d: Dict[str, int]):
    """
    Args:
        This is a method for calculating the algoritmic complexity and it received the following arguments
        L: A list of integers
        m: Another list of integers
        d: a dictionary with the same length than L and s
    """
    r = 0
    for k in d.keys():
        for i, j in zip(L, m):
            d[k] = i * j
            r = d[k] + 1
    return r
```

```python
def algoritmicComplexity(L, m, d):
    """
    Args:
        This is a method for calculating the algoritmic complexity and it received the following arguments
        L: A list of integers
        m: Another list of integers
        d: a dictionary with the same length than L and s
    """
    r = 0
    for k in d.keys():
        for i, j in zip(L, m):
            d[k] = i * j
            r = d[k] + 1
    return r
```

```python
import random
# importem random per generar numeros aleatoris per fer les llistes
# creem una llista on guardarem per el numero n tamany de les llistes quant hem tardat
total = []
t = 2000
# iterem per llistes de tamany 1 fins a 2000 en salts de 10
for n in range(1,t,10):
    # gennerem aleatoriament tant les llistes com el diccionari
    L = random.sample(range(1, t), n)
    m = random.sample(range(1, t), n)
    d = dict()
    d_keys = random.sample(range(1, t), n)
    for ele in d_keys:     
        # assigning random elements
        d[ele] = random.randint(1, t) 
    # calculem el temps que tardem per cada n i afegim a total quan tardem per una n i quina es la n
    a = %timeit -o -n 10 -r 5 algoritmicComplexity(L, m, d)
    total.append((n,np.mean(a.timings)))
```

```python
# mostrem graficament els temps que ha tardat per cada iteracio
import matplotlib.pyplot as plt

plt.scatter(*zip(*total))
```

<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px; ">SM</span> **(1.5 punts)**

```python

from random import randint


def count_appearence(iters, n):
    """
     Given a number n, generates a random list of 'iters'
     number of elements between [-10*n,10*n] and count how many times
     our input n appears in the random list.

     Args:
         iters (:obj: `int`) : integer, number of random elements to generate
         n (:obj: `int`) : integer, number to count appearences
     Returns:
         (:obj: `int`) : integer, number of times that n
         appears in the generated random list.
    """
    c = 0
    L = []
    for i in range(iters):
        L.append(randint(-10*n, 10*n))
    for j in L:
        if j == n:
            c += 1
    return c


```

```python

iters, n = int(1e6), int(1e2)
%time count_appearence(iters, n)
```

El primer que podem optimitzar es l'acces a la llista, tecnicament no ens importa masa tenir o no la llista ja que no la retornem, apart de que la estem recorrent dos vegades, al crearla i al compobar si els elements son iguals, per tant el que podem fer es generar iters nombres aleaotris i comprobar si son iguals en el mateix pas. Veiem que fent això es una mica més eficient però encara podem millorar mes

```python
def count_appearence2(iters, n):
    """
     Given a number n, generates a random list of 'iters'
     number of elements between [-10*n,10*n] and count how many times
     our input n appears in the random list.

     Args:
         iters (:obj: `int`) : integer, number of random elements to generate
         n (:obj: `int`) : integer, number to count appearences
     Returns:
         (:obj: `int`) : integer, number of times that n
         appears in the generated random list.
    """
    c = 0
    for i in range(iters):
        if randint(-10*n, 10*n) == n:
            c += 1
        
    return c
```

```python

iters, n = int(1e6), int(1e2)
%time count_appearence2(iters, n)
```

El que podem fer es generar iters/n nombres aleatoris entre -10,10 i mirar si son iguals a 1, podem fer això ja la distribucio es la mateixa.
ja que la possibilitat de que randint sigui igual a n es de 1/2n*10 i la posibilitat que entre -10 i 10 sigui 1 es de 1/20 per tant si fem iter a un i iter/n a l'altre la distribucui de valors hauria de ser la mateixa, i veiem que c es del mateix ordre i reduim molt la complexitat, casi 100 vegades això es degut a que estem generant nombres mes petits lo qual es mes facil i que tallem la iteracions per .
Per tant al principi creixia com iter*(dificultat de generar nombres aleatoris de -10n,10n) + iter( comprovacio)
el segon es iter*(dificultat de generar nombres aleatoris de -10n,10n + comprovacio)
i el tercer es (iter/n)*(dificultat de generar nombres aleatoris de -10,10 + comprovacio)

```python
def count_appearence3(iters, n):
    """
     Given a number n, generates a random list of 'iters'
     number of elements between [-10*n,10*n] and count how many times
     our input n appears in the random list.

     Args:
         iters (:obj: `int`) : integer, number of random elements to generate
         n (:obj: `int`) : integer, number to count appearences
     Returns:
         (:obj: `int`) : integer, number of times that n
         appears in the generated random list.
    """
    c = 0
    for i in range(int(iters/n)):
        if randint(-10, 10) == 1:
            c += 1
        
    return c
```

```python
%time count_appearence3(iters, n)
```

```python
def squaresum(n: int):
    """devuelve la suma de los cuadrados para los primeros n números naturales.

    Parámetros de entrada:
        n: número entero

    Valores que devuelve:
        (int): suma de cuadrados de 1 a n.
    """
    total = 0
    for i in range(1, n + 1):
        for j in range(i):  # Bucle innecesario para calcular el cuadrado
            total += i
    return total
```

**Resposta:**

Simplement utilitzar la opercacio elevar al 2 encomptes de sumar n vegades fa que el calcul sigui molt mes eficient
ja que pasem de fer una suma n^2 vegades a fer una exponencial n vegades i veiem que la diferencia en temps es molt notable

```python
def squaresum2(n: int):
    """devuelve la suma de los cuadrados para los primeros n números naturales.

    Parámetros de entrada:
        n: número entero

    Valores que devuelve:
        (int): suma de cuadrados de 1 a n.
    """
    total = 0
    for i in range(1, n + 1):
        total += i**2
    return total
```

## Exercici 5
<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px; ">SM</span> **(1 punt)**

En aquest exercici treballarem el multiprocessament: Es donen com donar-vos d'entrada el codi següent:

```python
#import List
import time
import multiprocessing
import math

# Lista de números grandes para calcular el factorial
numeros = [20000, 25000, 30000, 35000, 40000]

# Función para calcular el factorial de un número
def calcular_factorial(n: int):
    """Devuelve el factorial de un número:
    Los datos de entrada será el número a calcular,
    mientras que los datos que devuelve es el factorial del número ya calculado
    """
    return math.factorial(n)
```

### 5.1 (0.75 punt)
Es demana crear dues funcions que rebin com a paràmetre d'entrada la llista de números a calcular:
- La primera calcularà el factorial de la llista de manera seqüencial.
- La segona calcularà el factorial de la llista utilitzant el multiprocessament.

Ambdues funcions hauran de mostrar el temps que triguen a calcular la llista

**Solució 5.1**

```python
def secuencial(numbers):
    fact = []
    for i in numbers:
        fact.append(calcular_factorial(i))
    return fact
```

```python
%time factorials = secuencial(numeros)
```

```python
from multiprocessing import Process
import logging

def multiproces(numeros):
    processes = []
    for i in numeros:    
        # Crea el procés
        p = Process(target=calcular_factorial, args=(i,))
        processes.append(p)
        # Inicia el procés
        p.start()
    # Espera a que els 3 processos finalitzin
    for process in processes:
        process.join()
    return
    
```

```python
%time multiproces(numeros)
```

Quant de temps han trigat a cadascuna de les funcions?

## Exercici 6

Una biblioteca té un projecte on es tracta de recuperar informació de diversos llibres sobre la base de la seva [ISBN](https://en.wikipedia.org/wiki/International_Standard_Book_Number). En una primera instància ens demanen recuperar el títol del llibre, autor o autors i una imatge de portada per poder crear una pàgina web de consulta.

La biblioteca també es planteja contractar els serveis d'un proveïdor cloud per allotjar els serveis de consulta i visualització de la informació i els preocupa principalment el temps que trigarà el servei de consulta.

En una anàlisi prèvia, s'ha investigat l'ús de l'API de *[Open Library](https://openlibrary.org/developers/api)* per consultar la informació que se'ns demana. A continuació podeu veure un exemple d'ús d'aquesta API. La següent funció retorna el títol, autor/s i imatge de portada donat un ISBN.

```python
import requests


def get_book_info(isbn):
    """Returns title, author and cover image URL from
    Open Library given an ISBN.

    Parameters:
    isbn (string): ISBN to search.

    Returns:
    tuple: tuple in the form (title (string), authors (list),
            cover image url (string))
    """

    openLibraryUrl = "https://openlibrary.org/api/books"
    parameters = {
        "bibkeys": f"ISBN:{isbn}",
        "format": "json",
        "jscmd": "data",
    }
    response = requests.get(openLibraryUrl, params=parameters)
    book_info = response.json()[f"ISBN:{isbn}"]

    title = book_info["title"]
    authorNames = []
    for author in book_info["authors"]:
        authorNames.append(author["name"])
    coverImageUrl = (
        book_info["cover"]["small"] if "cover" in book_info else ""
    )

    return (title, authorNames, coverImageUrl)
```

Primer hauriem de ententre la diferencia entre multiprocesing i multitheard multithreading fa referència a la capacitat d'un processador d'executar múltiples fils simultàniament, on cada fil executa un procés.
El multiprocessament es refereix a la capacitat d'un sistema per executar diversos processadors en paral·lel, on cada processador pot executar un o més fils.
El multithreading és útil per als processos vinculats a IO, com ara llegir fitxers d'una xarxa o base de dades, ja que cada fil pot executar el procés vinculat a IO simultàniament. El multiprocessament és útil per als processos vinculats a la unitat central de processament (CPU), com ara tasques computacionalment pesades, ja que es beneficiarà de tenir diversos processadors; similar a com els ordinadors multinucli funcionen més ràpid que els ordinadors amb un sol nucli.
En el nostre cas depenem de les calls a la api per tant seria millor multithreading ja que el coll d'ampolla no estara limitat per la capacitat de comput si no per les calls a la api

```python
get_book_info(isbns_to_test[0])
```

```python
import concurrent.futures
def multiplethread(i):
    with concurrent.futures.ThreadPoolExecutor(max_workers=i) as executor:
        # Start the load operations and mark each future with its URL
        future_to_url = {executor.submit(get_book_info, isbns): isbns for isbns in isbns_to_test}
        for future in concurrent.futures.as_completed(future_to_url):
            url = future_to_url[future]
            try:
                data = future.result()
            except Exception as exc:
                print('%r generated an exception: %s' % (url, exc))
            else:
                print(data)
```

```python
%time multiplethread(5)
```

```python
for i in range(1,11):
    print(i)
    %time multiplethread(i)
```

Veiem com en general contra mes threads tenim més rapid va, en els casos que va mes lent pot ser perfectament culpa del servidor que tarda més en retornarnos resposta sobre la nostra consulta
