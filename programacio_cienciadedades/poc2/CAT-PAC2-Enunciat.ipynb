# Exercici 1

Volem crear una funció que comprovi si es pot realitzar la mescla entre ingredients d'una pizza. Els ingredients que pot introduir es classifiquen en dos tipus:

Principals:

- Tofu
- Bacon
- Embotit picant

Secundaris:

- Oliva
- Xampinyons
- Pinya

La funció rebrà dos paràmetres:
- Una llista amb 3 ingredients
- un paràmetre **opcional** que ens indiqui si volem que la funció mostri per pantalla la raó per la qual no es pot realitzar la pizza (en el cas que si es pugui, la funció no haurà de mostrar cap missatge de totes maneres).

La funció retornarà una dada de tipus booleà **(True / False)**.

Segons la pizzeria, els únics ingredients que no es poden posar en mateixa pizza són els següents:

- Bacon i Tofu excepte si se li afegeix oliva
- Bacon, xampinyons i Pinya
- Embotit picant i oliva excepte si s'afegeix xampinyons

Les combinacions han de ser obligatòriament de 3 ingredients però l'única combinació de 2 ingredients permesa és tofu amb oliva.

No es pot fer una mescla de més de 3 ingredients ni menys de 3, a excepció de la que s'ha comentat.

Com a segona part d'aquest exercici, **utilitzant programació funcional**,
realitza el següent:

- Quines d'aquestes combinacions d'ingredients es poden realitzar i quins no
- Mostrar la/s tupla/s dels ingredients que es poden realitzar.

<pre>
\[ ["Bacon", "Embotit picant", "Pinya"],["Oliva", "Pinya", "Xampinyons"],["Tofu", "Bacon", "Oliva", "Pinya"]  \]
</pre>

<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px; ">SM</span> **(1.5 punt)**

```python
def ingredients_valids(ingredients, output):
    '''Donada una llista d'ingredients, valida si la combinacio
    d'ingredients es correcte. Si es demana, també diu el motiu
    perque no s'acepta la comanda per pantalla. Retorna cert o fals
    depenent de si es valida o no'''
    # Primer comprovem si te 3 ingredients, ja que si te mes
    # o menys no es valida excepte en una excepcio
    if len(ingredients) == 3:
        # En cas que siguin tres mirem si es alguna de les
        # combinacions no valides amb la funció in per mirar
        # si a la llista hi ha cert elements
        if ("Bacon" in ingredients and
                "Tofu" in ingredients and "Oliva" not in ingredients):
            # En cas que output sigui cert mostrem
            # per pantalla el motiu de rebuitg
            if output:
                print("La pitza de Bacon i Tofu nomes pot portar Oliva")
            return False
        elif ("Bacon" in ingredients and
                "Xampinyons" in ingredients and "Pinya" in ingredients):
            if output:
                print("La pitza de Bacon , Xampinyons i Pinyma no es pot demanar")
            return False
        elif ("Embotit picant" in ingredients and
                "Oliva" in ingredients and "Xampinyons" not in ingredients):
            if output:
                print("La pitza de Embotit picant i Oliva nomes pot portar Xampinyons")
            return False
        return True
    else:
        # En cas que no tingui tres l'unica
        # valida es tofu i oliva per tant
        # comprovem si es aquesta igual que a dalt
        if len(ingredients) == 2:
            if ("Tofu" in ingredients and "Oliva" in ingredients):
                return True
        if output:
            print("El numero d'ingredients no es l'adequat, les pitzes nomes poden tenir 3 ingredients, exeptuant la pitza de dos ingredients de tofu amb oliva")
        return False
```

```python
comandes = [["Bacon", "Embotit picant", "Pinya"], ["Oliva", "Pinya", "Xampinyons"], ["Tofu", "Bacon", "Oliva", "Pinya"]]
# definim la llista
final_comandes = list(map(lambda x: ingredients_valids(x, False), comandes))
# iterem per la llista per aplicar la funcio a cada comanda individual
print(final_comandes)
# la mostrem per pantalla
# explicitem per pantalla quines comandes es poden demanar i quines no
for i in range(len(comandes)):
    if final_comandes[i]:
        print("la comanda "+str(comandes[i])+" es pot demanar")
    else:
        print("la comanda "+str(comandes[i])+" no es pot demanar")
final_comandes = list(filter(lambda x: ingredients_valids(x, False), comandes))
# utilitzant filter ens quedem amb les comandes valides
print(final_comandes)
```

```python
import pandas as pd
# llegim el excel i generem un df i creem una nova columna
# que dividint entre 1000 per pasar de m a km
volcans = pd.read_excel("data/volcanes.xlsx")
volcans = volcans.assign(elevation_km=lambda x: (x['Elevation (m)']/1000))
```

```python
def create_dictionary_with_elevation(path):
    """"donat una ruta on hi ha un excel llegim l'exel
    i generem un diccionari on la clau es el nom del volca i
    el valor la seva elevacio en meteres"""
    volcans = pd.read_excel(path)
    # llegim el excel com a df
    volcans_dict = volcans[["Volcano Name", "Elevation (m)"]].set_index('Volcano Name')["Elevation (m)"].to_dict()
    # utilitzant la funcio to_dict generem un diccionari
    # de les columnes del df que volem
    return (volcans_dict)
```

```python
volcan_dict = create_dictionary_with_elevation("data/volcanes.xlsx")
```

```python
from operator import itemgetter


def sort_top5(dict_to_sort):
    """ dontat un dicctionari ordenem els
    valors del elements i ens quedem amb els 5 primers"""
    # utilitzem sorted per ordenar el diccionari
    # i itemgetter per optenir el valor  dels elements del diccionari i
    # utilitzant [:5] ens quedem amb els 5 primers elements
    return dict(sorted(dict_to_sort.items(), key=itemgetter(1), reverse=True)[:5])
```

# Exercici 3

Es disposa d'un diccionari `vendes` d'un comerç, el qual disposa dels productes, que és la clau del diccionari, i la quantitat venuda (valor) al llarg d'un mes.

Per exemple:

vendes = {'aigua': 1000, 'llet': 300, 'sal': 1000, 'sucre': 350,'oli': 312}

També es disposa d'un diccionari *estoc_ini*, de manera que els items són els noms de productes (clau) i la quantitat del producte (valor) que que el comerç tenia a primers del mes.

Per exemple:

estoc_ini = {'sal':1000, 'oli':1000, 'llet':500, 'macarrons': 400, 'sucre':500, 'colacao': 1600, 'aigua': 1000}

Al final de mes s'ha d'actualitzar l'estoc d'acord amb el que s'ha venut.

Pels exemples anteriors el nou estoc seria:

estoc_final = {'sal': 0,
'oli': 688,
'llet':200,
'macarrons': 400,
'sucre':150,
'colacao': 1600,
'aigua': 0}

Considerarem que que mai s'ha venut més quantitats d'un producte del que hi havia en estoc.

Dissenya la funció `inventari(vendes, estoc_ini)` que calculi i mostri l'estoc final segons l'especificació i a més que mostri un missatge que indiqui que indiqui quins procuctes són els que s'han de reposar.

<span style="font-family: Courier New; background-color: #ffcc5c; color: #000000; padding: 3px; ">SM</span> **(1.5 punts)**



```python
%pycodestyle_off
# Definim els dos diccionearis
vendes = {'aigua': 1000, 'llet': 300, 'sal': 1000, 'sucre': 350, 'oli': 312}
estoc_ini = {'sal':1000, 'oli':1000, 'llet':500, 'macarrons': 400, 'sucre':500, 'colacao': 1600, 'aigua': 1000}
%pycodestyle_on
```

```python
def inventari(vendes, estoc_ini):
    """donats dos diccionaris  un de ventes
    i un d'estoc calculem quin es l'estoc final restant els elements"""
    estoc_final = estoc_ini
    # iterem per les claus de vendes i calculem la resta
    # de l'estoc final menys el de ventes
    for i in vendes.keys():
        estoc_final[i] = estoc_ini[i]-vendes[i]
        # si algun estoc esta a 0 avisem per pantalla quin es
        if estoc_final[i] == 0:
            print("S'ha acabat l'estoc del producte "+str(i))
    return (estoc_final)
```

```python
estoc_final = inventari(vendes, estoc_ini)
print(estoc_final)
```

# Exercici 4

Es demana realitzar una funció que recorri l'excel noms.xlsx:

La funció que es sol·licita s'anomenarà `random_asignation`, rebrà com a entrada:

- L'arxiu de dades en format *.xlsx*.

- Un número comprès entre 1 i 245.

La sortida serà un diccionari on:

- La clau serà el número corresponent a la posició, és a dir, ANTONIO, que és el primer nom, tindrà de clau un 1.

- Una tupla que contindrà el nom i un número aleatori comprès entre 1 i 100 que haureu de crear.

Per acabar, amb el número que s'ha sol·licitat al principi, es crearà aquesta quantitat de números aleatoris i correspondran a les claus que es mostraran com a sortida final de la funció. És a dir, si m'han indicat un 3 al principi, haig de generar 3 números aleatoris, en aquest cas m'han aparegut: 1, 5 i 10. La funció haurà de retornar:

<pre>
\[('ANTONIO', 20), ('DAVID', 7),('JOSE ANTONIO', 21)\]
</pre>

Recordeu que els números que apareixen al costat del nom són aleatoris, i per tant no us apareixeran els mateixos números que en aquest exemple.

<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px; ">EG</span> **(2.0 punts)**

```python
import pandas as pd
import random


def random_asignation(path, num_of):
    """donada una ruta i un nombre generem una llista
    de n valors on n es el numero generat, cada valor
    sera un nom que treurerm d'un excel  que esta a la
    ruta donada i un numero aleatori del 1 al 100.
    Per decidir quins elements agafem per la llista final
    generarem n numeros aleatoris i agafarem l'element en
    la possicio del numero aleatori"""
    list_to_return = []
    # definim una llista
    noms = pd.read_excel(path)
    # generem un df de la llista
    name_dict = noms["Nombres"].to_dict()
    # el pasem a diccionari
    # generem la llista de "num_of"
    # parelles on cada element es l'element i del diccionari
    # conjuntament amb un numero aleatori, on i es també un numero aleatori
    for i in range(num_of):
        list_to_return.append((name_dict[random.randint(1, 100)], random.randint(1, 100)))
    return (list_to_return)
```

# Exercici 5

Per a aquest exercici, primer s'haurà de crear una funció per a extreure la informació de l'arxiu .zip

Primer de tot es demana crear una funció `zip_extraction`que rebrà el nom del fitxer i el nom de la carpeta on es vol descomprimir. El zip a extreure es diu `Jorge_Luis.zip`. En aquest cas, l'arxiu .zip conté fitxers de text i que la cerca de la paraula es realitza sobre aquests fitxers.

Per a la segona part es demana crear una funció `analyse_zip` que rebi el nom de l'arxiu zip i una tupla amb les paraules que es vol que es busquin en tots els arxius que es trobin en el zip.

El resultat que haurà de mostrar la funció és la suma de totes les vegades que es troben les paraules que s'indiquen en la tupla.

**Notes importants:**

1) Per a la segona funció es pot incloure la funció `analyse_zip` dins per a poder descomprimir l'arxiu en cas que el necessiti.

2) Utilitzeu els principis de *programació funcional* que hem vist en teoria per a resoldre aquest exercici.

<span style="font-family: Courier New; background-color: #ffcc5c; color: #000000; padding: 3px; ">EG</span> **(2.5 punts)**

```python
import zipfile as zf


def zip_extraction(path_file, path_end=""):
    """donat una ruta d'entrada i
    una de sortida descomprimim un arxiu d'una ruta a l'atre"""
    with zf.ZipFile(zip_file, 'r') as zip_f:
        # Descomprimim tot el contingut del zip a la ruta path_end
        zip_f.extractall(path_end)
```

```python
def analyse_zip(path_file, paraules):
    """ donat una ruta i una llista de paraules,
    busquem a la ruta un arxiu zip i mirem dintre de la
    carpeta comprimida tots els arxius '.txt' i contem
    quantes vegades surten les paraules de la llista
    de paraules a els arxius '.txt.'"""
    # generem una variable on posarem el numero
    # total de vegades que es diuen les paraules
    total = 0
    with zf.ZipFile(path_file, 'r', metadata_encoding="utf-8") as zip_f:
        # obrim l'arxiu comprimit a la ruta donada i guardem
        # a una llista tots els arxius ".txt" dintre de la carpeta comprimida
        achivos = [x for x in zip_f.infolist() if ".txt" in x.filename]
        for j in archivos:
            # per cada arxiu l'obrim i el decodejem perque utilitzi "uft8"
            # per accents i altres caracters i el pasem a str. despres
            # iterem per la llista de paraules i utilitzant la funcio "count()"
            # de str contem quantes vegades surt cada parula a cada text
            with zip_f.open(j.filename) as myfile:
                text = str(myfile.read().decode('utf8'))
                for i in paraules:
                    total += text.count(i)
    return (total)
```

```python

paraules=["dia","la","Junín,"]
path_file="data/Jorge_Luis.zip"
analyse_zip(path_file, paraules)
```
